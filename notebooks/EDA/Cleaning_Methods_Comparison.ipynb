```python
# Block 2: Imports
import sys
import os
import pandas as pd
import numpy as np
import warnings
from pathlib import Path
import shutil

warnings.filterwarnings("ignore")

sys.path.append(os.path.abspath("."))

from src.dataset.loader import download_from_drive, read_csv_with_columns
from src.dataset.cleaning import (
    remove_html_and_entities,
    remove_urls,
    remove_mentions,
    remove_emoji,
    remove_special_chars,
    remove_null_or_empty,
    remove_non_text_comments,
    remove_duplicate_comments,
    remove_outliers,
)
from src.dataset.feature_enginering import (
    add_length_features,
    add_punctuation_emoji_features,
)
from src.dataset.preprocessing import normalize_text
from config.dataset_config import default_dataset_config
from config.cleaning_config import default_cleaning_config

print("Import successful!")
```

```python
# Block 4: Read original dataset
df_original = read_csv_with_columns(
    data_path=LOCAL_PATH,
    comment_col=COMMENT_COL,
    label_col=LABEL_COL,
)

print(f"\nOriginal dataset shape: {df_original.shape}")
print(f"Columns: {df_original.columns.tolist()}")
```

---
## 9 Cleaning Methods

Each block below creates 1 dataset applying **only 1 method**, keeping everything else unchanged.

Datasets are saved directly to Google Drive in the same folder as the original file.

```python
# ============================================================
# Method 1: remove_html_and_entities
# Remove HTML tags and decode HTML entities
# ============================================================
print("=" * 60)
print("Method 1: remove_html_and_entities")
print("=" * 60)

df = df_original.copy()
df[COMMENT_COL] = df[COMMENT_COL].astype(str).apply(remove_html_and_entities)

output_name = "dataset_cleaned_01_remove_html_and_entities.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 2: remove_urls
# Remove URLs (http, https, ftp, www)
# ============================================================
print("=" * 60)
print("Method 2: remove_urls")
print("=" * 60)

df = df_original.copy()
df[COMMENT_COL] = df[COMMENT_COL].astype(str).apply(remove_urls)

output_name = "dataset_cleaned_02_remove_urls.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 3: remove_mentions
# Remove @mentions
# ============================================================
print("=" * 60)
print("Method 3: remove_mentions")
print("=" * 60)

df = df_original.copy()
df[COMMENT_COL] = df[COMMENT_COL].astype(str).apply(remove_mentions)

output_name = "dataset_cleaned_03_remove_mentions.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 4: remove_emoji
# Remove emoji
# ============================================================
print("=" * 60)
print("Method 4: remove_emoji")
print("=" * 60)

df = df_original.copy()
df[COMMENT_COL] = df[COMMENT_COL].astype(str).apply(remove_emoji)

output_name = "dataset_cleaned_04_remove_emoji.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 5: remove_special_chars
# Keep only letters, numbers, spaces, and punctuation (default .,!?)
# ============================================================
print("=" * 60)
print("Method 5: remove_special_chars")
print("=" * 60)

df = df_original.copy()
df[COMMENT_COL] = (
    df[COMMENT_COL]
    .astype(str)
    .apply(lambda x: remove_special_chars(x, keep_punctuation=default_cleaning_config.keep_punctuation))
)

output_name = "dataset_cleaned_05_remove_special_chars.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 6: remove_null_or_empty
# Remove rows with null/empty/whitespace-only comments, remove null labels
# ============================================================
print("=" * 60)
print("Method 6: remove_null_or_empty")
print("=" * 60)

df = df_original.copy()
df, report = remove_null_or_empty(
    df,
    comment_col=COMMENT_COL,
    label_col=LABEL_COL,
    max_null_label_ratio=default_cleaning_config.max_null_label_ratio,
)

output_name = "dataset_cleaned_06_remove_null_or_empty.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Removed {report['empty_comment_removed']} empty comments")
print(f"Removed {report['null_label_removed']} null label rows")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 7: remove_non_text_comments
# Remove comments without any letter (e.g. "12345", "!!!", "emoji")
# ============================================================
print("=" * 60)
print("Method 7: remove_non_text_comments")
print("=" * 60)

df = df_original.copy()
df, report = remove_non_text_comments(
    df,
    comment_col=COMMENT_COL,
)

output_name = "dataset_cleaned_07_remove_non_text_comments.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Removed {report['non_text_removed']} non-text comments")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 8: remove_duplicate_comments
# Same comment + same label -> keep first; same comment + different label -> remove all
# ============================================================
print("=" * 60)
print("Method 8: remove_duplicate_comments")
print("=" * 60)

df = df_original.copy()
df, report = remove_duplicate_comments(
    df,
    comment_col=COMMENT_COL,
    label_col=LABEL_COL,
)

output_name = "dataset_cleaned_08_remove_duplicate_comments.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Duplicate groups: {report['duplicate_comment_groups']}")
print(f"Removed (different labels): {report['removed_rows_different_labels']}")
print(f"Removed (same label): {report['removed_rows_duplicate_same_label']}")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 9: remove_outliers
# Detect and remove outliers using Isolation Forest
# NOTE: This method requires numeric feature columns.
# We use add_length_features and add_punctuation_emoji_features
# from src/dataset/feature_enginering.py to create them.
# ============================================================
print("=" * 60)
print("Method 9: remove_outliers (Isolation Forest)")
print("=" * 60)

df = df_original.copy()

# Create numeric features using predefined functions from feature_enginering.py
df = add_length_features(df, comment_col=COMMENT_COL)
df = add_punctuation_emoji_features(df, comment_col=COMMENT_COL)

# Apply remove_outliers
df, report = remove_outliers(
    df,
    feature_cols=default_cleaning_config.outlier_feature_cols,
    contamination=default_cleaning_config.outlier_contamination,
    random_state=default_cleaning_config.outlier_random_state,
    label_col=LABEL_COL,
)

# Drop temporary feature columns before saving
df = df.drop(
    columns=["word_len", "char_len", "num_exclamation", "num_question", "num_upper", "num_emoji"],
    errors="ignore",
)

output_name = "dataset_cleaned_09_remove_outliers.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"Shape: {df.shape}")
print(f"Removed {report['outlier_count']} outliers ({report['outlier_ratio']:.2%})")
print(f"Saved: {output_path}")
print()
```

```python
# ============================================================
# Method 10: ALL methods combined (full cleaning pipeline)
# Apply every cleaning method in a logical order:
#   1. Text-level: html -> urls -> mentions -> emoji -> special_chars
#   2. Dataset-level: null/empty -> non-text -> duplicates -> outliers
# ============================================================
print("=" * 60)
print("Method 10: ALL cleaning methods combined")
print("=" * 60)

df = df_original.copy()
initial_rows = len(df)

# ---- Step 1: Text-level transformations ----
print("\n[Step 1/4] remove_html_and_entities...")
df[COMMENT_COL] = df[COMMENT_COL].astype(str).apply(remove_html_and_entities)

print("[Step 1/4] remove_urls...")
df[COMMENT_COL] = df[COMMENT_COL].apply(remove_urls)

print("[Step 1/4] remove_mentions...")
df[COMMENT_COL] = df[COMMENT_COL].apply(remove_mentions)

print("[Step 1/4] remove_emoji...")
df[COMMENT_COL] = df[COMMENT_COL].apply(remove_emoji)

print("[Step 1/4] remove_special_chars...")
df[COMMENT_COL] = df[COMMENT_COL].apply(
    lambda x: remove_special_chars(x, keep_punctuation=default_cleaning_config.keep_punctuation)
)

# ---- Step 2: Remove null/empty rows ----
print("\n[Step 2/4] remove_null_or_empty...")
df, report_null = remove_null_or_empty(
    df,
    comment_col=COMMENT_COL,
    label_col=LABEL_COL,
    max_null_label_ratio=default_cleaning_config.max_null_label_ratio,
)
print(f"  Removed {report_null['empty_comment_removed']} empty comments")
print(f"  Removed {report_null['null_label_removed']} null label rows")

# ---- Step 3: Remove non-text comments ----
print("\n[Step 3/4] remove_non_text_comments...")
df, report_non_text = remove_non_text_comments(
    df,
    comment_col=COMMENT_COL,
)
print(f"  Removed {report_non_text['non_text_removed']} non-text comments")

# ---- Step 4: Remove duplicates ----
print("\n[Step 4/4] remove_duplicate_comments...")
df, report_dup = remove_duplicate_comments(
    df,
    comment_col=COMMENT_COL,
    label_col=LABEL_COL,
)
print(f"  Duplicate groups: {report_dup['duplicate_comment_groups']}")
print(f"  Removed (different labels): {report_dup['removed_rows_different_labels']}")
print(f"  Removed (same label): {report_dup['removed_rows_duplicate_same_label']}")

# ---- Step 5: Remove outliers ----
print("\n[Step 5/5] remove_outliers (Isolation Forest)...")
df_temp = add_length_features(df, comment_col=COMMENT_COL)
df_temp = add_punctuation_emoji_features(df_temp, comment_col=COMMENT_COL)

df, report_out = remove_outliers(
    df_temp,
    feature_cols=default_cleaning_config.outlier_feature_cols,
    contamination=default_cleaning_config.outlier_contamination,
    random_state=default_cleaning_config.outlier_random_state,
    label_col=LABEL_COL,
)

# Drop temporary feature columns
df = df.drop(
    columns=["word_len", "char_len", "num_exclamation", "num_question", "num_upper", "num_emoji"],
    errors="ignore",
)
print(f"  Removed {report_out['outlier_count']} outliers ({report_out['outlier_ratio']:.2%})")

# ---- Summary ----
print("\n" + "=" * 60)
print("FINAL RESULT")
print("=" * 60)
print(f"Initial rows: {initial_rows}")
print(f"Final rows:   {len(df)}")
print(f"Removed:      {initial_rows - len(df)} rows")

output_name = "dataset_cleaned_10_all_methods_combined.csv"
output_path = os.path.join(DRIVE_OUTPUT_DIR, output_name)
df.to_csv(output_path, index=False, encoding="utf-8-sig")

print(f"\nSaved: {output_path}")
print()
```
